```python
import pandas as pd 
```

```python
df = pd.read_csv("student_scores.csv")
df.head(10)
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=["Score"])
```

```python
df.isnull().sum()
```

```python
df["Grade"] = df["Score"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)
```

```python
df.head(10)
```

```python
df.duplicated().sum()
```

```python
Top_Students = df[df["Score"] >= 90]

Top_Students
```

```python
Top_Students[["StudentID", "Name", "Subject", "Score", "Grade"]]
```

```python
len(Top_Students)
```

```python
average_score = df.groupby("Subject")["Score"].mean().sort_values(ascending=False)
average_score
```

```python
gender_score = df.groupby("Gender")["Score"].agg(["mean","min","max"])
gender_score
```

```python
subject_count = df["Subject"].value_counts()
subject_count
```

```python
df.to_csv("cleaned_students.csv,index=False")
print("Cleaned dataset exported successfully!")
```

```python
df.head(10)
```

```python
subject_summary = df.groupby("Subject")["Score"].mean()
subject_summary
```

```python
subject_summary.to_excel("subject_summary.xlsx")

print("Subject summary exported successfully!")
```
